## Data Cleaning Decisions

The following cleaning rules were applied to prepare the dataset for customer-level analysis and segmentation:

1. Exact duplicate transaction records were removed.
2. Transactions with invoice numbers beginning with `C` were removed because the dataset documentation identifies these as cancellations.
3. Transactions with non-positive quantities were excluded from the purchase dataset.
4. Transactions with non-positive unit prices were excluded.
5. Records with missing `CustomerID` were excluded because customer-level segmentation requires an identifiable customer.
6. Missing product descriptions were retained because `StockCode` remains available and the description is not required for the planned RFM analysis.
7. `InvoiceDate` was retained as a datetime field for subsequent recency calculations.

These decisions were made specifically to create a reliable transaction-level dataset for customer behavior analysis and machine-learning-based customer segmentation.
