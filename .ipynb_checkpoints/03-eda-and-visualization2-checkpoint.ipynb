```python
clean_df = pd.read_csv('clean_online_retail.csv')
```

```python
clean_df.head()
```

## Overall Business Overview

```python
print("Revenue = Quantity x UnitPrice")
clean_df["Revenue"] = clean_df["Quantity"] * clean_df["UnitPrice"] 

clean_df[["Quantity",  "UnitPrice", "Revenue"]].head()
```

```python
clean_df.Revenue.describe()
```

```python
total_revenue = clean_df.Revenue.sum()
print("Total Revenue:", total_revenue)

print("Total number of transactions:", clean_df.InvoiceNo.nunique())
print("Total number of customers:", clean_df.CustomerID.nunique())
print("Total number of products:", clean_df.StockCode.nunique())
print("Total number of countries:", clean_df.Country.nunique())
```

```python
clean_df["InvoiceDate"] = pd.to_datetime(clean_df["InvoiceDate"])
```

## Revenue Overtime

```python
clean_df["Month"] = clean_df["InvoiceDate"].dt.to_period("M")
```

```python
clean_df.head()
```

## Revenue Over Time
Track total revenue by month to see business trends

```python
# Calculate monthly revenue
monthly_revenue = clean_df.groupby('Month')['Revenue'].sum()

# Plot
fig, ax = plt.subplots(figsize=(14, 5))
monthly_revenue.plot(ax=ax, marker='o', linewidth=2, markersize=6, color='#2E86AB')
ax.set_title('Total Revenue by Month', fontsize=14, fontweight='bold')
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Revenue (£)', fontsize=12)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Highest revenue month: {monthly_revenue.idxmax()} (£{monthly_revenue.max():,.2f})")
print(f"Lowest revenue month: {monthly_revenue.idxmin()} (£{monthly_revenue.min():,.2f})")
```

##  Number of Orders Over Time
How many transactions are happening each month?

```python
# Calculate monthly order count
monthly_orders = clean_df.groupby('Month')['InvoiceNo'].nunique()

# Plot
fig, ax = plt.subplots(figsize=(14, 5))
monthly_orders.plot(ax=ax, marker='s', linewidth=2, markersize=6, color='#A23B72')
ax.set_title('Number of Orders by Month', fontsize=14, fontweight='bold')
ax.set_xlabel('Month', fontsize=12)
ax.set_ylabel('Number of Orders', fontsize=12)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Peak order month: {monthly_orders.idxmax()} ({monthly_orders.max()} orders)")
print(f"Lowest order month: {monthly_orders.idxmin()} ({monthly_orders.min()} orders)")
```

## Revenue by Country
Which countries are generating the most revenue?

```python
# Calculate revenue by country
country_revenue = clean_df.groupby('Country')['Revenue'].sum().sort_values(ascending=False).head(15)

# Plot
fig, ax = plt.subplots(figsize=(12, 8))
country_revenue.plot(kind='barh', ax=ax, color='#F18F01')
ax.set_title('Top 15 Countries by Revenue', fontsize=14, fontweight='bold')
ax.set_xlabel('Revenue (£)', fontsize=12)
ax.set_ylabel('Country', fontsize=12)
plt.tight_layout()
plt.show()

print(f"\nTop 5 Countries:")
print(country_revenue.head())
print(f"\nTop country (UK) revenue: £{country_revenue.iloc[0]:,.2f}")
```

##  Top 15 Products by Revenue
Which products are our revenue drivers?

```python
# Calculate revenue by product (StockCode)
product_revenue = clean_df.groupby('StockCode')['Revenue'].sum().sort_values(ascending=False).head(15)

# Plot
fig, ax = plt.subplots(figsize=(12, 8))
product_revenue.plot(kind='barh', ax=ax, color='#06A77D')
ax.set_title('Top 15 Products by Revenue', fontsize=14, fontweight='bold')
ax.set_xlabel('Revenue (£)', fontsize=12)
ax.set_ylabel('Product Code', fontsize=12)
plt.tight_layout()
plt.show()

print(f"\nTop 5 Products:")
print(product_revenue.head())
```

## Customer Spending Distribution
How much does each customer spend?

```python
# Calculate total spending per customer
customer_spending = clean_df.groupby('CustomerID')['Revenue'].sum()

# Create subplots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
axes[0].hist(customer_spending, bins=50, color='#2E86AB', edgecolor='black', alpha=0.7)
axes[0].set_title('Distribution of Customer Spending', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Total Spending (£)', fontsize=11)
axes[0].set_ylabel('Number of Customers', fontsize=11)
axes[0].grid(True, alpha=0.3, axis='y')

# Box plot
axes[1].boxplot(customer_spending, vert=True)
axes[1].set_title('Customer Spending - Box Plot', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Total Spending (£)', fontsize=11)
axes[1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

print(f"Customer Spending Statistics:")
print(customer_spending.describe())
print(f"\nHighest spending customer: £{customer_spending.max():,.2f}")
print(f"Lowest spending customer: £{customer_spending.min():,.2f}")
```

## Purchase Frequency
How often do customers buy from us?

```python
# Calculate number of purchases per customer
purchase_frequency = clean_df.groupby('CustomerID')['InvoiceNo'].nunique()

# Create subplots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
axes[0].hist(purchase_frequency, bins=50, color='#A23B72', edgecolor='black', alpha=0.7)
axes[0].set_title('Distribution of Purchase Frequency', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Number of Purchases', fontsize=11)
axes[0].set_ylabel('Number of Customers', fontsize=11)
axes[0].grid(True, alpha=0.3, axis='y')

# Top customers by frequency
top_freq_customers = purchase_frequency.nlargest(10)
axes[1].barh(range(len(top_freq_customers)), top_freq_customers.values, color='#F18F01')
axes[1].set_yticks(range(len(top_freq_customers)))
axes[1].set_yticklabels(top_freq_customers.index)
axes[1].set_title('Top 10 Most Frequent Customers', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Number of Purchases', fontsize=11)
axes[1].grid(True, alpha=0.3, axis='x')

plt.tight_layout()
plt.show()

print(f"Purchase Frequency Statistics:")
print(purchase_frequency.describe())
print(f"\nMost frequent customer: {purchase_frequency.idxmax()} ({purchase_frequency.max()} purchases)")
```

## Quantity Analysis
How many items are sold per order?

```python
# Quantity statistics
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram of quantity per transaction
axes[0].hist(clean_df['Quantity'], bins=50, color='#06A77D', edgecolor='black', alpha=0.7)
axes[0].set_title('Distribution of Quantity per Transaction', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Quantity', fontsize=11)
axes[0].set_ylabel('Frequency', fontsize=11)
axes[0].grid(True, alpha=0.3, axis='y')

# Total quantity sold over time
monthly_quantity = clean_df.groupby('Month')['Quantity'].sum()
axes[1].plot(monthly_quantity.index.astype(str), monthly_quantity.values, 
             marker='o', linewidth=2, markersize=6, color='#2E86AB')
axes[1].set_title('Total Quantity Sold by Month', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Month', fontsize=11)
axes[1].set_ylabel('Total Quantity', fontsize=11)
axes[1].grid(True, alpha=0.3)
plt.setp(axes[1].xaxis.get_majorticklabels(), rotation=45)

plt.tight_layout()
plt.show()

print(f"Quantity Statistics:")
print(clean_df['Quantity'].describe())
```

## Unit Price Analysis
What's the price distribution of our products?

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram of unit prices
axes[0].hist(clean_df['UnitPrice'], bins=50, color='#A23B72', edgecolor='black', alpha=0.7)
axes[0].set_title('Distribution of Unit Prices', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Unit Price (£)', fontsize=11)
axes[0].set_ylabel('Frequency', fontsize=11)
axes[0].grid(True, alpha=0.3, axis='y')

# Log scale histogram (for better visibility)
axes[1].hist(clean_df['UnitPrice'], bins=50, color='#F18F01', edgecolor='black', alpha=0.7)
axes[1].set_title('Distribution of Unit Prices (Log Scale)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Unit Price (£)', fontsize=11)
axes[1].set_ylabel('Frequency (log)', fontsize=11)
axes[1].set_yscale('log')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Unit Price Statistics:")
print(clean_df['UnitPrice'].describe())
print(f"\nPrice range: £{clean_df['UnitPrice'].min():.2f} - £{clean_df['UnitPrice'].max():.2f}")
```

##  Revenue Contribution Analysis
Which customer segments and products drive revenue?

```python
# Top 10 customers by revenue contribution
top_customers = clean_df.groupby('CustomerID')['Revenue'].sum().nlargest(10)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Top customers
axes[0].barh(range(len(top_customers)), top_customers.values, color='#2E86AB')
axes[0].set_yticks(range(len(top_customers)))
axes[0].set_yticklabels(top_customers.index)
axes[0].set_title('Top 10 Customers by Revenue', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Revenue (£)', fontsize=11)
axes[0].grid(True, alpha=0.3, axis='x')

# Revenue contribution pie chart for top countries
top_countries = clean_df.groupby('Country')['Revenue'].sum().nlargest(8)
other_revenue = clean_df.groupby('Country')['Revenue'].sum().nsmallest(clean_df['Country'].nunique() - 10).sum()
plot_data = pd.concat([top_countries, pd.Series({'Other': other_revenue})])

colors = ['#2E86AB', '#A23B72', '#F18F01', '#06A77D', '#C73E1D', '#6A994E', '#BC4749', '#2E86AB', '#F4D35E']
axes[1].pie(plot_data, labels=plot_data.index, autopct='%1.1f%%', colors=colors[:len(plot_data)], startangle=90)
axes[1].set_title('Revenue Distribution by Country', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

print(f"\nTop 10 Customers Revenue Contribution:")
print(top_customers)
total_revenue = clean_df['Revenue'].sum()
print(f"\nTop 10 customers account for: {(top_customers.sum()/total_revenue)*100:.1f}% of total revenue")
```

## Key Insights Summary
What did we learn from this EDA?

```python
print("="*70)
print("KEY INSIGHTS FROM EDA")
print("="*70)

total_revenue = clean_df['Revenue'].sum()
total_customers = clean_df['CustomerID'].nunique()
total_transactions = clean_df['InvoiceNo'].nunique()
avg_order_value = total_revenue / total_transactions
avg_customer_value = total_revenue / total_customers

print(f"\n BUSINESS METRICS:")
print(f"   Total Revenue: £{total_revenue:,.2f}")
print(f"   Total Transactions: {total_transactions:,}")
print(f"   Total Customers: {total_customers:,}")
print(f"   Average Order Value: £{avg_order_value:,.2f}")
print(f"   Average Customer Value: £{avg_customer_value:,.2f}")
print(f"   Products in Catalog: {clean_df['StockCode'].nunique():,}")
print(f"   Countries Served: {clean_df['Country'].nunique()}")

print(f"\n REVENUE TRENDS:")
monthly_revenue = clean_df.groupby('Month')['Revenue'].sum()
print(f"   Peak month: {monthly_revenue.idxmax()} (£{monthly_revenue.max():,.2f})")
print(f"   Lowest month: {monthly_revenue.idxmin()} (£{monthly_revenue.min():,.2f})")

print(f"\n GEOGRAPHIC INSIGHTS:")
top_country = clean_df.groupby('Country')['Revenue'].sum().idxmax()
top_country_revenue = clean_df.groupby('Country')['Revenue'].sum().max()
print(f"   Top market: {top_country} (£{top_country_revenue:,.2f})")
print(f"   Market concentration: Top country accounts for {(top_country_revenue/total_revenue)*100:.1f}% of revenue")

print(f"\n CUSTOMER INSIGHTS:")
customer_spending = clean_df.groupby('CustomerID')['Revenue'].sum()
print(f"   Average customer spending: £{customer_spending.mean():,.2f}")
print(f"   Median customer spending: £{customer_spending.median():,.2f}")
print(f"   Highest spender: £{customer_spending.max():,.2f}")


print("="*70)
```
